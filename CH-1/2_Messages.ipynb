# **Messages**

```python
from langchain_google_genai import ChatGoogleGenerativeAI

from dotenv import load_dotenv
import os
load_dotenv()


if os.environ['GEMINI_API_KEY']:
    print('gemini key set')
else:
    print("not set")
```

## **Initialize LLM**

```python
# Initialize the Gemini model
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

my_messages = [
    HumanMessage(content="How are you?"),
    SystemMessage(content="You're a gen-z assistant who answers in a fun playful way")
]

llm.invoke(my_messages).content[0]["text"]
```

## **Prompts**

prompts are more user friendly than messages

```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter topic about fact")

dynamic_prompt = PromptTemplate.from_template("write a fun fact about: {topic}")

ready_prompt = dynamic_prompt.invoke({"topic":user_input})

llm.invoke(ready_prompt).content[0]["text"]
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system","you are a helpful assistant"),
    ("user","Write a fun fact about {topic}")
])

chain = prompt_template | llm

user_insert = input("Enter a topic to get fun fact")

chain.invoke({"topic":user_insert}).content[0]["text"]
```
