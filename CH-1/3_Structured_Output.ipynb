# **Guiding in Prompts**

```python
from langchain_google_genai import ChatGoogleGenerativeAI

from dotenv import load_dotenv
import os
load_dotenv()


if os.environ['GEMINI_API_KEY']:
    print('gemini key set')
else:
    print("not set")
```

## **Initialize LLM**

```python
# Initialize the Gemini model
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
)
```

```python
from langchain_core.prompts import PromptTemplate

llm.invoke("Tell me a joke. Generate the output in key-value pair format with the following keys: setup, punchline.").content[0]['text']
```

# **Using Pydantic Models**

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup: str = Field(description="setup for the joke")
    punchline: str = Field(description="punchline for the joke")
```

```python
obj = llm_schema(**{"setup":"some setup","punchline":"some punchline"})
obj.setup
```

```python
llm_structured_output = llm.with_structured_output(llm_schema)
result = llm_structured_output.invoke("Tell me a joke about corn")
result
```

# **Using TypeDict**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup: str
    punchline: str
```

```python
obj_td = llm_schema_td({"setup":"some setup","punchline":"some punchline"})
obj_td.get('setup')
```

```python
llm_structured_type_dict = llm.with_structured_output(llm_schema_td)

llm_structured_type_dict.invoke("Tell me a joke about avocado")
```
