```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.runnables import RunnableSequence, RunnableLambda

from dotenv import load_dotenv
import os
load_dotenv()


if os.environ['GEMINI_API_KEY']:
    print('gemini key set')
else:
    print("not set")
```

## **Chain With Custom Runnable**

```python

# Task 1, [Prompt]
prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant"),
    ("user", "{input}")
])
```

```python
# Task 2, LLM

llm_task = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
)
```

```python
# Task 3, [String Parser]
str_parser = StrOutputParser()
```

```python
# Task 4, [Custom Runnable]

def dictionary_maker(text:str) -> dict:
    return {"text":text}


dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
# Task 5 [Template for post]

prompt_post = ChatPromptTemplate.from_messages([
    ("system","You are a social media post generator"),
    ("human","Create a post for the following text for twitter(X): {text}"),
]
)
```

```python
# Chain invocation
chain = prompt_template | llm_task | str_parser | dictionary_maker_runnable | prompt_post | llm_task | str_parser

chain.invoke({"input":"Fun fact about bees"})
```
